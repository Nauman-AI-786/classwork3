```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
```

```python
# house prediction dataset
x= np.array([[1], [2], [3], [4], [5]]) # 1 dimensional array / area
y= np.array([2, 4, 6, 8, 10]) # 1 dimensional array / price

model = LinearRegression()
model.fit(x, y) #independent variable and dependent variable

prediction= model.predict(np.array([[6]])) # predict the price of house with area 6
print("predicted price of house with area 6:", prediction[0])
```

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

x = np.array([[6,2], [4,3], [6,5], [7,8], [5,6]])
y = np.array([0,0,1,1,1]) # 0 fail, 1 = pass

model = KNeighborsClassifier(n_neighbors=3)
model.fit(x,y)

prediction= model.predict(np.array([[6, 4]])) #predict the result of student with marks 5 and 4

print("Predicted result of student with marks 6 and 4:", prediction)
 
```

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

#x = np.array([[Math, English, Biology ]])
x = np.array([[6, 2, 5],  #student 1
    [4, 3, 4], # Stundent 2
    [6, 5, 7],  #studnet 3
    [7, 8, 9],   #student 4
    [5, 6, 6]     #studnet 5
    ])

y = np.array([0,0,1,1,1])   # 0=fail 1= pass

# model initialize aur train karein
model = KNeighborsClassifier(n_neighbors=3)

model.fit(x,y)


# Naye student ke 3 subjects ke marks ke sath predict karein (Misal: [6, 4, 5])
new_student = np.array([[6, 4, 5]])
prediction = model.predict(new_student)

print("Predicted result (0=Fail, 1=Pass):", prediction)

```

```python
import pandas as pd

# Aapka data (3 subjects + 1 target result/pass-fail)
data = {
    'math': [6, 4, 6, 7, 5],
    'english': [2, 3, 5, 8, 6],
    'biology': [5, 4, 7, 9, 6],
    'result': [0, 0, 1, 1, 1]  # 0 = Fail, 1 = Pass
}

# DataFrame banayein
df = pd.DataFrame(data)

# CSV file ke tor par save karein
df.to_csv('dataclass.csv', index=False)

print("Mubarak ho! 'dataclass.csv' file ban gayi hai.")
```
